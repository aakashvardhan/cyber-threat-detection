#### based on HF documentation

```python
state = "Hi, we were billed twice for March. Please refund the duplicate today or we will cancel our plan."
questions = {
    "department": {"type": "choice", "instructions": "Which department should handle this?",
                   "criteria": {"billing": "invoices, payments, refunds",
                                "technical": "bugs, outages, system errors",
                                "other": "everything else"}},
    "urgency": {"type": "score", "instructions": "How urgent is this?",
                "criteria": ["not urgent", "soon", "blocking"]},
    "churn_risk": {"type": "noul", "instructions": "Does the user threaten to cancel or leave?"},
}
```

```python

result = router.predict(state, questions)
print(result["answers"]["department"]["choice"])  # billing
print(result["answers"]["churn_risk"]["noul"])    # probability the answer is yes
print(result["routing"]["model"])                 # english
```

**testing all three tasks**

```python
tactics = [
    "Reconnaissance",
    "Discovery",
    "Credential Access",
    "Resource Development",
    "Privilege Escalation",
    "Defense Evasion",
    "Initial Access",
    "Persistence",
    "Exfiltration",
    "Lateral Movement",
    "Execution",
    "Command and Control",
    "Collection",
]

techniques = [
    "T1046", "T1595", "T1110", "T1587",
    "T1018", "T1592", "T1590", "T1016",
    "T1548", "T1078", "T1190", "T1048",
    "T1210", "T1589", "T1203", "T1071",
    "T1566", "T1204", "T1059", "T1547",
    "T1136", "T1112", "T1571", "T1557",
    "T1546", "T1505", "T1133",
]

questions = {
    "detection": {
        "type": "choice",
        "instructions": (
            "Classify the target connection as attack or benign "
            "using the supplied traffic features and context."
        ),
        "criteria": {
            "1": "Attack",
            "0": "Benign",
        },
    }
}

for tactic in tactics:
    questions[f"tactic_{tactic}"] = {
        "type": "choice",
        "instructions": (
            f"Does the target connection exhibit the MITRE ATT&CK "
            f"tactic '{tactic}'? Evaluate this independently; "
            "multiple tactics may apply."
        ),
        "criteria": {
            "yes": f"The connection exhibits {tactic}.",
            "no": f"The connection does not exhibit {tactic}.",
        },
    }

for technique in techniques:
    questions[f"mitre_{technique}"] = {
        "type": "choice",
        "instructions": (
            f"Does the target connection exhibit MITRE ATT&CK "
            f"technique {technique}? Evaluate this independently; "
            "multiple techniques may apply."
        ),
        "criteria": {
            "yes": f"The connection exhibits {technique}.",
            "no": f"The connection does not exhibit {technique}.",
        },
    }

print("Total questions:", len(questions))  # 41
```

```python
answers = result.get("answers", result)

binary = str(answers["detection"]["choice"])

tactics = [
    key.removeprefix("tactic_")
    for key, answer in answers.items()
    if key.startswith("tactic_")
    and str(answer["choice"]).lower() == "yes"
]

techniques = [
    key.removeprefix("mitre_")
    for key, answer in answers.items()
    if key.startswith("mitre_")
    and str(answer["choice"]).lower() == "yes"
]

print("Binary:", {"1": "Attack", "0": "Benign"}.get(binary, binary))
print("Techniques:", techniques)
print("Tactics:", tactics)
```
